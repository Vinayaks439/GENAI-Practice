# Stage 2 — Data Collection

Download the Mahabharata corpus from Kaggle and stage it locally under `data/raw/`.

```python
import kagglehub

path = kagglehub.dataset_download('tilakd/mahabharata')
print('Path to dataset files:', path)
```

```python
import os
from pathlib import Path

src = Path(path)
for f in sorted(src.rglob('*')):
    if f.is_file():
        size_kb = f.stat().st_size / 1024
        print(f'{size_kb:>10.1f} KB  {f.relative_to(src)}')
```

```python
import shutil

raw_dir = Path('data/raw')
raw_dir.mkdir(parents=True, exist_ok=True)

for f in src.rglob('*'):
    if f.is_file():
        dest = raw_dir / f.name
        if not dest.exists():
            shutil.copy2(f, dest)

print('Staged files:')
for f in sorted(raw_dir.iterdir()):
    print(' ', f.name, f'({f.stat().st_size / 1024:.1f} KB)')
```

```python
# Peek at the largest text-like file.
text_files = [p for p in raw_dir.iterdir() if p.suffix.lower() in {'.txt', '.csv'}]
text_files.sort(key=lambda p: p.stat().st_size, reverse=True)

for p in text_files[:3]:
    print(f'\n=== {p.name} ===')
    with p.open('r', encoding='utf-8', errors='replace') as fh:
        head = fh.read(800)
    print(head)
```
